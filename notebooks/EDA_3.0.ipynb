计算相关性矩阵

```python
print("\n" + "=" * 80)
print("2. 相关性矩阵计算")
print("=" * 80)

corr_matrix = df.corr()
print(f"\n相关性矩阵大小: {corr_matrix.shape}")
```

与目标变量的相关性（完整排序）

```python
print("\n" + "=" * 80)
print("3. 与 Diabetes_012 的相关性分析")
print("=" * 80)

# 获取完整的相关性（按实际值排序，非绝对值）
target_corr = corr_matrix["Diabetes_012"].sort_values(ascending=False)

print("\n完整相关性排序 (ALL FEATURES):")
print("-" * 80)
for i, (feat, corr_val) in enumerate(target_corr.items(), 1):
    if feat != 'Diabetes_012':
        direction = "Positive (Risk Factor)" if corr_val > 0 else "Negative (Protective Factor)"
        strength = "Strong" if abs(corr_val) > 0.2 else "Moderate" if abs(corr_val) > 0.1 else "Weak"
        print(f"  {i:2d}. {feat:25s}: r = {corr_val:7.4f}  [{strength:8s}, {direction}]")
```

保存结果到 DataFrame

```python
print("\n" + "=" * 80)
print("4. 保存相关性数据")
print("=" * 80)

# 创建 DataFrame
correlation_df = pd.DataFrame({
    'Feature': target_corr.index,
    'Correlation': target_corr.values,
    'Abs_Correlation': target_corr.abs().values,
    'Direction': ['Risk' if x > 0 else 'Protective' for x in target_corr.values],
    'Strength': ['Strong' if abs(x) > 0.2 else 'Moderate' if abs(x) > 0.1 else 'Weak' 
                 for x in target_corr.values]
}).query('Feature != "Diabetes_012"')

# 保存到CSV
correlation_df.to_csv('feature_correlations.csv', index=False)
print("✓ 已保存: feature_correlations.csv")

# 显示数据
print("\n前10个特征:")
print(correlation_df.head(10).to_string(index=False))
```

## Log Transformation Impact Analysis for BMI

**Rationale for Exclusion**: Demonstrating why Log Transformation is inappropriate for BMI in this public health context.

```python
# Create comparison visualization: BMI before and after Log Transformation
# Highlighting the distance compression in the 70-100 high-risk interval

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Define colors for risk zones
colors_original = ['#2ecc71', '#f39c12', '#e74c3c']  # Green, Orange, Red
colors_log = ['#3498db', '#9b59b6']  # Blue, Purple

# Plot 1: Original BMI Distribution (Right-skewed with long tail)
ax1 = axes[0]
n, bins, patches = ax1.hist(df['BMI'], bins=50, edgecolor='black', linewidth=0.5, alpha=0.7)

# Color code the bins by risk level
for i, patch in enumerate(patches):
    if bins[i] < 25:
        patch.set_facecolor(colors_original[0])  # Normal weight
    elif bins[i] < 30:
        patch.set_facecolor(colors_original[1])  # Overweight
    else:
        patch.set_facecolor(colors_original[2])  # Obese

ax1.axvline(70, color='darkred', linestyle='--', linewidth=2.5, label='High-Risk Threshold (70)', alpha=0.8)
ax1.axvline(100, color='darkred', linestyle='--', linewidth=2.5, alpha=0.8)

# Annotate the physical distance in original scale
ax1.annotate('', xy=(100, ax1.get_ylim()[1]*0.8), xytext=(70, ax1.get_ylim()[1]*0.8),
            arrowprops=dict(arrowstyle='<->', color='darkred', lw=3))
ax1.text(85, ax1.get_ylim()[1]*0.85, 'Physical Distance:\n30 kg/m² gap', 
         ha='center', fontsize=11, fontweight='bold', color='darkred',
         bbox=dict(boxstyle='round', facecolor='yellow', alpha=0.7))

ax1.set_xlabel('BMI (Original Scale)', fontweight='bold', fontsize=12)
ax1.set_ylabel('Frequency', fontweight='bold', fontsize=12)
ax1.set_title('Figure A: Original BMI Distribution\n(Preserves Critical High-Risk Distances)', 
             fontweight='bold', fontsize=13, pad=15)
ax1.legend(loc='upper right', fontsize=10)
ax1.grid(axis='y', alpha=0.3)

# Add annotation for extreme values
extreme_count = (df['BMI'] >= 70).sum()
ax1.text(0.98, 0.5, f'Extreme BMI ≥70:\n{extreme_count:,} cases\n({extreme_count/len(df)*100:.2f}%)',
        transform=ax1.transAxes, ha='right', va='center',
        bbox=dict(boxstyle='round', facecolor='white', edgecolor='red', alpha=0.9),
        fontsize=10, fontweight='bold')

# Plot 2: Log-Transformed BMI Distribution (More symmetric but compressed)
ax2 = axes[1]
bmi_log = np.log(df['BMI'])
n_log, bins_log, patches_log = ax2.hist(bmi_log, bins=50, edgecolor='black', linewidth=0.5, 
                                        alpha=0.7, color='#9b59b6')

# Mark the same 70 and 100 thresholds in log scale
log_70 = np.log(70)
log_100 = np.log(100)
ax2.axvline(log_70, color='darkred', linestyle='--', linewidth=2.5, label='log(70) = 4.25', alpha=0.8)
ax2.axvline(log_100, color='darkred', linestyle='--', linewidth=2.5, alpha=0.8)

# Annotate the compressed distance in log scale
ax2.annotate('', xy=(log_100, ax2.get_ylim()[1]*0.8), xytext=(log_70, ax2.get_ylim()[1]*0.8),
            arrowprops=dict(arrowstyle='<->', color='darkred', lw=3))
ax2.text((log_70+log_100)/2, ax2.get_ylim()[1]*0.85, 
         'Compressed Distance:\nOnly 0.36 log units!\n(Was 30 kg/m²)', 
         ha='center', fontsize=11, fontweight='bold', color='darkred',
         bbox=dict(boxstyle='round', facecolor='yellow', alpha=0.7))

ax2.set_xlabel('log(BMI) - Logarithmic Scale', fontweight='bold', fontsize=12)
ax2.set_ylabel('Frequency', fontweight='bold', fontsize=12)
ax2.set_title('Figure B: Log-Transformed BMI Distribution\n⚠️ PROBLEM: High-Risk Distances Artificially Compressed', 
             fontweight='bold', fontsize=13, pad=15, color='#c0392b')
ax2.legend(loc='upper left', fontsize=10)
ax2.grid(axis='y', alpha=0.3)

# Add warning box
ax2.text(0.02, 0.5, '⚠️ CLINICAL RISK:\nBMI 70→100 represents\na 43% increase in mass.\n\nLog transformation\nreduces this to just\n8.5% visual difference,\nmasking critical risk\nescalation!',
        transform=ax2.transAxes, ha='left', va='center',
        bbox=dict(boxstyle='round', facecolor='#ffcccc', edgecolor='darkred', linewidth=2, alpha=0.95),
        fontsize=10, fontweight='bold', color='darkred')

plt.tight_layout()
plt.savefig('bmi_log_transformation_comparison.png', dpi=300, bbox_inches='tight')
print("\n" + "="*80)
print("✓ 已保存: bmi_log_transformation_comparison.png")
print("="*80)
plt.show()

# Quantitative Analysis
print("\n" + "="*80)
print("QUANTITATIVE EVIDENCE: Why Log Transformation Fails for BMI")
print("="*80)

# Calculate actual distances
bmi_70_to_100_original = 100 - 70
bmi_70_to_100_log = np.log(100) - np.log(70)

print(f"\n1. ORIGINAL SCALE:")
print(f"   Distance between BMI 70 and 100: {bmi_70_to_100_original} kg/m²")
print(f"   Physical meaning: {(100/70 - 1)*100:.1f}% increase in body mass")

print(f"\n2. LOG SCALE:")
print(f"   Distance between log(70) and log(100): {bmi_70_to_100_log:.4f} log units")
print(f"   Visual compression ratio: {(bmi_70_to_100_original / bmi_70_to_100_log):.1f}x")

# Compare with lower BMI range
bmi_20_to_30_original = 30 - 20
bmi_20_to_30_log = np.log(30) - np.log(20)

print(f"\n3. COMPARISON WITH NORMAL RANGE (BMI 20-30):")
print(f"   Original scale distance: {bmi_20_to_30_original} kg/m²")
print(f"   Log scale distance: {bmi_20_to_30_log:.4f} log units")
print(f"   Log transformation makes extreme range (70-100) appear")
print(f"   {bmi_20_to_30_log / bmi_70_to_100_log:.2f}x MORE similar than normal range!")

print(f"\n4. CLINICAL IMPLICATION:")
high_risk_count = ((df['BMI'] >= 70) & (df['BMI'] <= 100)).sum()
print(f"   Number of patients in 70-100 BMI range: {high_risk_count:,}")
print(f"   Percentage of total population: {high_risk_count/len(df)*100:.2f}%")
print(f"   ⚠️  These {high_risk_count:,} highest-risk individuals would be")
print(f"   mathematically 'flattened' by log transformation, making them")
print(f"   indistinguishable from moderately obese patients (BMI 35-50).")

print("\n" + "="*80)
print("CONCLUSION: Linear Robust Scaling is the only appropriate choice")
print("to preserve clinical risk gradients for vulnerable population analysis.")
print("="*80)
```
